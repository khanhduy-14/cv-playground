```python
import os
import matplotlib.pyplot as plt

from PIL import Image

print("CWD: " + os.getcwd())
img = Image.open("../../assets/threshold/cat.jpg").convert('RGB')

plt.imshow(img)
plt.axis('off')
plt.show()

```

```python
import numpy as np

img_array = np.array(img)

R = img_array[:, :, 0]
G = img_array[:, :, 1]
B = img_array[:, :, 2]


gray = 0.299 * R + 0.587 * G + 0.114 * B

gray = gray.astype(np.uint8)



plt.imshow(gray, cmap='gray')
plt.axis('off')
plt.show()
```

```python
counts = np.bincount(gray.flatten(), minlength=256)
intensities = np.arange(256)

plt.figure(figsize=(12, 4))
plt.bar(intensities, counts, width=1.0, color='black')
plt.xlim(0, 255)


plt.xlabel('Intensity Value')
plt.ylabel('Pixel Count')
plt.title('Histogram of Grayscale Image')
plt.grid(axis='y', alpha=0.3)
plt.show(   )
```

```python
threshold = 210
gray[gray>threshold] = 255
gray[gray<=threshold] = 0


fig, axes = plt.subplots(1,2 ,figsize=(10,5))

axes[0].imshow(img)
axes[0].set_title('Original cat image')

axes[1].imshow(gray, cmap='gray')
axes[1].set_title('Segmented image by threshold based')

plt.show()
```
